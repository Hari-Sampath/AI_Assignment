```python
import math


class TicTacToe:

    def __init__(self):
        self.board = [
            ['X', 'O', 'X'],
            ['O', 'X', ' '],
            [' ', ' ', 'O']
        ]

    def print_board(self):
        for row in self.board:
            print(row)

    def available_moves(self):

        moves = []

        for i in range(3):
            for j in range(3):

                if self.board[i][j] == ' ':
                    moves.append((i, j))

        return moves

    def make_move(self, move, player):
        i, j = move
        self.board[i][j] = player

    def undo_move(self, move):
        i, j = move
        self.board[i][j] = ' '

    def winner(self):

        lines = []

        lines.extend(self.board)

        for col in range(3):
            lines.append([self.board[row][col] for row in range(3)])

        lines.append([self.board[i][i] for i in range(3)])
        lines.append([self.board[i][2 - i] for i in range(3)])

        for line in lines:

            if line == ['X', 'X', 'X']:
                return 'X'

            if line == ['O', 'O', 'O']:
                return 'O'

        if len(self.available_moves()) == 0:
            return 'Draw'

        return None


def alpha_beta(game, alpha, beta, is_maximizing):

    result = game.winner()

    if result == 'X':
        return 1

    elif result == 'O':
        return -1

    elif result == 'Draw':
        return 0

    if is_maximizing:

        value = -math.inf

        for move in game.available_moves():

            game.make_move(move, 'X')

            value = max(value,
                        alpha_beta(game, alpha, beta, False))

            game.undo_move(move)

            alpha = max(alpha, value)

            if alpha >= beta:
                break

        return value

    else:

        value = math.inf

        for move in game.available_moves():

            game.make_move(move, 'O')

            value = min(value,
                        alpha_beta(game, alpha, beta, True))

            game.undo_move(move)

            beta = min(beta, value)

            if beta <= alpha:
                break

        return value


def best_move(game):

    best_score = -math.inf
    best_move_choice = None

    for move in game.available_moves():

        game.make_move(move, 'X')

        score = alpha_beta(game,
                           -math.inf,
                           math.inf,
                           False)

        game.undo_move(move)

        if score > best_score:
            best_score = score
            best_move_choice = move

    return best_move_choice


# =========================
# TEST CASE
# =========================

game = TicTacToe()

print("Current Board:")
game.print_board()

move = best_move(game)

print("\nBest Move using Alpha-Beta:", move)
```
